```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.preprocessing import StandardScaler


df = pd.read_csv('/content/construction_project_forecasting_data.csv')


X = df[['work_progress', 'worker_availability', 'material_availability',
        'equipment_availability', 'delays', 'resource_shortage', 'budget_overrun']]
y_completion = df['predicted_completion_days_left'] 
y_workers = df['additional_workers_needed']
y_materials = df['additional_materials_needed']  


X_train, X_test, y_train_completion, y_test_completion = train_test_split(X, y_completion, test_size=0.2, random_state=42)
X_train, X_test, y_train_workers, y_test_workers = train_test_split(X, y_workers, test_size=0.2, random_state=42)
X_train, X_test, y_train_materials, y_test_materials = train_test_split(X, y_materials, test_size=0.2, random_state=42)


rf_completion = RandomForestRegressor(n_estimators=100, random_state=42)
rf_completion.fit(X_train, y_train_completion)


rf_workers = RandomForestRegressor(n_estimators=100, random_state=42)
rf_workers.fit(X_train, y_train_workers)


rf_materials = RandomForestRegressor(n_estimators=100, random_state=42)
rf_materials.fit(X_train, y_train_materials)


predictions_completion = rf_completion.predict(X_test)
predictions_workers = rf_workers.predict(X_test)
predictions_materials = rf_materials.predict(X_test)


mae_completion = mean_absolute_error(y_test_completion, predictions_completion)
mse_completion = mean_squared_error(y_test_completion, predictions_completion)

mae_workers = mean_absolute_error(y_test_workers, predictions_workers)
mse_workers = mean_squared_error(y_test_workers, predictions_workers)

mae_materials = mean_absolute_error(y_test_materials, predictions_materials)
mse_materials = mean_squared_error(y_test_materials, predictions_materials)


print(f"Completion Prediction - MAE: {mae_completion}, MSE: {mse_completion}")
print(f"Workers Prediction - MAE: {mae_workers}, MSE: {mse_workers}")
print(f"Materials Prediction - MAE: {mae_materials}, MSE: {mse_materials}")


example_input = [[70, 30, 85, 90, 15, 0, 10]] 
predicted_completion = rf_completion.predict(example_input)
predicted_workers = rf_workers.predict(example_input)
predicted_materials = rf_materials.predict(example_input)

print(f"Predicted Completion Days Left: {predicted_completion[0]}")
print(f"Predicted Additional Workers Needed: {predicted_workers[0]}")
print(f"Predicted Additional Materials Needed: {predicted_materials[0]}")
```

```python
import joblib
joblib.dump(rf_completion, 'model_predicted_completion.pkl')
joblib.dump(rf_workers, 'model_additional_workers.pkl')
joblib.dump(rf_materials, 'model_additional_materials.pkl')
print("Models saved successfully as .pkl files.")
```
